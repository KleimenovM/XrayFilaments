# Particle density evolution in the steady state free-streaming regime

In this notebook I investigate the evolution of particle population being injected constantly at $z=0$ and propagating along the GMF line, coinsiding with the $z$-axis.

To describe the process analytically, we have to account for the following effects:

* Synchrotron cooling

* Inverse Compton (IC) cooling

* Synchrotron radiation (angle-dependent)

* Inverse Compton radiation (angle-independent)

We start from cooling, calculate the steady-state picture for the electron distribution, and then come to calculation of photon emission in x-ray (synchrotron) and gamma-rays (IC)

## Steady-state electron density as a function of distance

#### Does pitch-angle change during propagation?

Pitch angle $\alpha$ is defined as $\sin\alpha = v_\perp / c = \beta_\perp$;

$\beta_\perp + \beta_\parallel = 1$

Total electron energy $E$ is given by $E = \gamma m c^2 = (\gamma_\perp + \gamma_\parallel)mc^2$

$\gamma$ and $\beta$ can be connected by a Lorentz boost parameter $\chi$: $\beta = \th\chi$, and $\gamma = (1-\beta^2)^{-1/2} = \left(\frac{\ch^2\chi}{\ch^2\chi - \sh^2\chi}\right)^{1/2} = \ch\chi$

We can compare energy and angle dissipation rates:

$ \dot\beta = \partial_t(\th\chi) = \dfrac{1}{\ch^2\chi}\dot{\chi} = \dfrac{\partial_t\ch\chi}{\ch^3\chi\th\chi} = \dfrac{(mc^2)^2\dot E}{E^3 \beta}$

Hence, we see that $\beta$ changes much slower than energy. Hence, the zero-order approach would be to assume angles stay constant along the GMF line and thus avoid the transport term in pitch-angle space.

### Solving the transport equation

With the notes above, as the propagation velocity remains constant for all particles, whose pitch angle is assumed fixed, we write the following transport equation

$$ d_t n(z, E, t) = -\partial_E(\dot E \,n(z,E,t)) + Q(z, E, t), $$

which in the steady-state case becomes even simpler. We also use $z = v_\parallel t$:

$$ v_\parallel \partial_z n(z,E) = -n(z,E)\,(\partial_E \dot{E})- \dot{E}\,\partial_E n(z,E) + Q(z,E,t)$$

As usual, we introduce the $s$ variable such as the latter is equivalent to

$$ d_s n(z(s), E(s)) = -n(z,E)\,(\partial_E \dot{E}) + Q(z,E,t)$$

The LHS is

$$(\partial_z n)\, \dfrac{dz}{ds} + (\partial_E n)\,\dfrac{dE}{ds} = n(z,E)\,(\partial_E \dot{E}) $$

Direct comparison with the initial equation gives us $ \dfrac{dz}{ds} = v_\parallel,~\dfrac{dE}{ds} = \dot{E}(s). $

The 1st equation has a straightforward solution: $s = z/v_\parallel$.

For the 2nd one, by setting the condition $E(s=0) = E_i$, we can choose a single characteristic.

Below, we will use $\dot{E} = \dot{E}_\mathrm{synch} + \dot{E}_\mathrm{IC}$



With our new variable $s = z/v_\parallel$, the equation became an ODE:

$ \dfrac{d}{ds} n(s) = -n(s) \partial_E \dot{E}(s) + Q(s), $ where $f(s) = \partial_E \dot{E}(s)$.

This equation can be solved *explicitly*:

$ \displaystyle n(s) = n_0\exp\left\{ - \int_{0}^{s} ds'\, \partial_E \dot{E}(s)(s') \right\} + \int_{0}^{s} ds' \exp\left\{-\int_{s'}^{s} ds''\,\partial_E \dot{E}(s'')\right\} $

As at z=0 there were initially no particles, we write 

$$ n(z, E) = \dfrac{1}{v_\parallel}\int_{0}^{z} dz'\,Q(E_i(z', E), z') \exp\left\{ -\dfrac{1}{v_\parallel}\int_{z'}^{z} dz'' \partial_E \dot{E} \right\} $$

Now we can use the assumption of particles being injected only at the source, thus writing

$ Q(E_i(z, E), z') = q(E)\delta(z) $

This removes one of the integrals, giving

$$ n(z, E) = \dfrac{q(E)}{v_\parallel} \exp\left\{ -\dfrac{1}{v_\parallel}\int_{0}^{z} dz' \partial_E \dot{E} (E_i(z', E)) \right\} $$

We are interested in the final pitch-angle distribution, which is defined by initial (honogeneuous) distribution
$q_\Omega = \dfrac{dN}{d\Omega\,dt} = \dfrac{1}{4\pi}q_0$. 

Or, which is even more relevant: $q_\alpha = \dfrac{dN}{d\cos\alpha\,dt} = q_\Omega\times\dfrac{d\Omega}{d\cos\alpha} = \dfrac{q_0}{2}$

Finally, $\displaystyle\dfrac{dN}{d\cos\alpha}(z, E) = \dfrac{q_\mathrm{tot}}{2c\cos\alpha}\exp\left\{ -\dfrac{1}{c\cos\alpha}\int_{0}^{z} dz' \partial_E \dot{E} [E_i(z', E), \alpha] \right\}  $

### Backward evolution of energy

First, we are to solve the evolution equation for energy.

To do that, we establish a "final" energy grid and z-grid. For fixed $\alpha$, we reconstruct "initial" energy from the final values by solving the following implicit ODE

$$ \dfrac{dE}{dz} = \dfrac{1}{c\cos\alpha}\dot{E}(E, \alpha) $$
