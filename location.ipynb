```python
def bar_aligned_ellipsis(a, angle=27 * u.deg, ratio=1/4):
    t = np.linspace(0, 2 * np.pi, 200)
    b = ratio * a
    bar_x1 = a * np.sin(t)
    bar_y1 = b * np.cos(t)
    bar_x = bar_x1 * np.cos(angle) + bar_y1 * np.sin(angle)
    bar_y = - bar_x1 * np.sin(angle) + bar_y1 * np.cos(angle)
    return bar_x, bar_y

def bar_rot(galc_coord, Omega_bar = 40):
    vel_vec = np.array([galc_coord.y.value, -galc_coord.x.value, .0]) * Omega_bar
    return vel_vec
```
